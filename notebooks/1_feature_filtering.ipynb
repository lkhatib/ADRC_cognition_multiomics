# Feature filtering

**Purpose:** Filter metagenomic features using genome coverage estimates generated by micov.

**Expected inputs**
- `../data/raw_feature_table.biom`
- `../output_files/coverage.tsv`

**Main outputs**
- `../data/micov_filtered_feature-table.biom`

```python
from pathlib import Path

PROJECT_ROOT = Path('..').resolve()
DATA_DIR = PROJECT_ROOT / 'data'
OUTPUT_DIR = PROJECT_ROOT / 'output_files'
FIGURE_DIR = PROJECT_ROOT / 'figures'

for directory in [DATA_DIR, OUTPUT_DIR, FIGURE_DIR]:
    directory.mkdir(parents=True, exist_ok=True)
```

```python
import pandas as pd
import qiime2 as q2
import biom
from biom import Table
```

```python
# Feature table from Qiita analysis 75191
qiita_table = biom.load_table("../data/75191_analysis_Metagenomic_Woltkav017pairedendDatabasescratchqpwoltkaWoLr2WoLr2BIOMnonebiomNA.biom")
```

## Drop features with lower than 1% coverage


```python
# Import coverages
cov = pd.read_csv('../output_files/coverage.tsv', sep = '\t')
```

```python
# Check how many features will be retained with a 1% threshold

len(cov[cov['percent_covered'] > 1])
```

```python
qiita_table = qiita_table.filter(cov[cov['percent_covered'] > 1]['genome_id'], axis='observation')
```

```python
cov.loc[cov['genome_id'].isin(qiita_table.to_dataframe().index)]
```

```python
with open("../data/micov_filtered_feature-table.biom", "w") as f:
    f.write(qiita_table.to_json("Generated by script"))
```
